```python
import os
import shutil

pasta_origem = r"D:\inmagem\New folder (2)\PhotoData\Metadata\PhotoData\CPLAssets"
pasta_destino = os.path.join(pasta_origem, "all")

# Criar pasta destino se não existir
os.makedirs(pasta_destino, exist_ok=True)

for raiz, dirs, arquivos in os.walk(pasta_origem):
    # Evita entrar na própria pasta "all"
    if raiz == pasta_destino:
        continue
    
    for arquivo in arquivos:
        caminho_origem = os.path.join(raiz, arquivo)
        caminho_destino = os.path.join(pasta_destino, arquivo)

        # Evita sobrescrever arquivos com mesmo nome
        if os.path.exists(caminho_destino):
            nome, ext = os.path.splitext(arquivo)
            contador = 1
            while os.path.exists(caminho_destino):
                novo_nome = f"{nome}_{contador}{ext}"
                caminho_destino = os.path.join(pasta_destino, novo_nome)
                contador += 1

        shutil.move(caminho_origem, caminho_destino)

print("Todos os arquivos foram movidos para a pasta 'all'")
```
